```python
def print_content(plots, era):
    print("Raw counts")
    print("QCD: ", plots[f"QCD_Pt_MuEnrichedPt5_{era}"]["raw"].values()[0])
    print("Data: ", plots[f"Data_{era}"]["raw"].values()[0])
    print(
        "QCD/Data: ",
        plots[f"Data_{era}"]["raw"].values()[0]
        / plots[f"QCD_Pt_MuEnrichedPt5_{era}"]["raw"].values()[0],
    )
    print(50 * "–")
    print()
    print("HLT_TripleMu_5_3_3_DZ_Mass3p8 counts")
    print(
        "QCD: ",
        plots[f"QCD_Pt_MuEnrichedPt5_{era}"]["HLT_TripleMu_5_3_3_DZ_Mass3p8"].values()[
            0
        ],
    )
    print("Data: ", plots[f"Data_{era}"]["HLT_TripleMu_5_3_3_DZ_Mass3p8"].values()[0])
    print(
        "QCD/Data: ",
        plots[f"Data_{era}"]["HLT_TripleMu_5_3_3_DZ_Mass3p8"].values()[0]
        / plots[f"QCD_Pt_MuEnrichedPt5_{era}"][
            "HLT_TripleMu_5_3_3_DZ_Mass3p8"
        ].values()[0],
    )
    print(50 * "–")
    print()
    print("final counts")
    print("QCD: ", plots[f"QCD_Pt_MuEnrichedPt5_{era}"]["final"].values()[0])
    print("Data: ", plots[f"Data_{era}"]["final"].values()[0])
    print(
        "QCD/Data: ",
        plots[f"Data_{era}"]["final"].values()[0]
        / plots[f"QCD_Pt_MuEnrichedPt5_{era}"]["final"].values()[0],
    )
```

```python
era = "2016"
plots = plot_utils.loader(tag=f"test_norm_May2025_{era}", era=era, load_data=True)
print_content(plots, era)
```

```python
def print_year_ratio(plots_2016, plots_2017):
    print("Raw counts")
    print(
        "2017/2016 QCD: ",
        round(
            plots_2017[f"QCD_Pt_MuEnrichedPt5_2017"]["raw"].values()[0]
            / plots_2016[f"QCD_Pt_MuEnrichedPt5_2016"]["raw"].values()[0],
            2,
        ),
    )
    print(
        "2017/2016 Data: ",
        round(
            plots_2017[f"Data_2017"]["raw"].values()[0]
            / plots_2016[f"Data_2016"]["raw"].values()[0],
            2,
        ),
    )
    print(50 * "–")
    print()
    print("HLT_TripleMu_12_10_5 counts")
    print(
        "2017/2016 QCD: ",
        round(
            plots_2017[f"QCD_Pt_MuEnrichedPt5_2017"]["HLT_TripleMu_12_10_5"].values()[0]
            / plots_2016[f"QCD_Pt_MuEnrichedPt5_2016"]["HLT_TripleMu_12_10_5"].values()[
                0
            ],
            2,
        ),
    )
    print(
        "2017/2016 Data: ",
        round(
            plots_2017[f"Data_2017"]["HLT_TripleMu_12_10_5"].values()[0]
            / plots_2016[f"Data_2016"]["HLT_TripleMu_12_10_5"].values()[0],
            2,
        ),
    )
    print(50 * "–")
    print()
    print("final counts")
    print(
        "2017/2016 QCD: ",
        round(
            plots_2017[f"QCD_Pt_MuEnrichedPt5_2017"]["processed"].values()[0]
            / plots_2016[f"QCD_Pt_MuEnrichedPt5_2016"]["processed"].values()[0],
            2,
        ),
    )
    print(
        "2017/2016 Data: ",
        round(
            plots_2017[f"Data_2017"]["processed"].values()[0]
            / plots_2016[f"Data_2016"]["processed"].values()[0],
            2,
        ),
    )
```
