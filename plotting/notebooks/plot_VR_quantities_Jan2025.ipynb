## nMuon distribution

```python
mc_processes = [
    "Higgs_2018",
    "ST_NLO_2018",
    "WJets_2018",
    "VV+VVV_2018",
    "TT_powheg_2018",
    "DY_2018",
    "QCD_Pt_MuEnrichedPt5_2018",
]

signal_processes = [
    "GluGluToSUEP_mS125.000_mPhi8.000_T8.000_modeleptonic_13TeV_2018",
    "GluGluToSUEP_mS125.000_mPhi4.000_T16.000_modeleptonic_13TeV_2018",
    "GluGluToSUEP_mS125.000_mPhi8.000_T16.000_modeleptonic_13TeV_2018",
    "GluGluToSUEP_mS125.000_mPhi8.000_T32.000_modeleptonic_13TeV_2018",
    "GluGluToSUEP_mS200.000_mPhi8.000_T32.000_modeleptonic_13TeV_2018",
]

hists_mc = []
hist_bkg_total = None
for process in mc_processes:
    h_mc = plots[process]["dimuon_mass_hard_vs_nMuon"][::sum, :]
    hists_mc.append(h_mc)
    if hist_bkg_total is None:
        hist_bkg_total = h_mc.copy()
    else:
        hist_bkg_total += h_mc.copy()

hists_signal = []
for process in signal_processes:
    h_s = plots[process]["dimuon_mass_hard_vs_nMuon"][::sum, :]
    hists_signal.append(h_s)

fig, ax = plt.subplots(figsize=(14, 13))

hep.histplot(
    hists_mc,
    yerr=[np.sqrt(h.variances()) for h in hists_mc],
    label=[label.replace("_2018", "") for label in mc_processes],
    stack=True,
    histtype="fill",
    ec="black",
    lw=2,
    ax=ax,
)

hep.histplot(
    hists_signal,
    yerr=[np.sqrt(h.variances()) for h in hists_signal],
    label=[
        s.replace("GluGluTo", "")
        .replace(".000", "")
        .replace("mode", "")
        .replace("_13TeV_2018", "")
        for s in signal_processes
    ],
    lw=3,
    ax=ax,
)

hep.cms.label(llabel="Preliminary", data=True, lumi=55, ax=ax)
plt.gca().xaxis.set_minor_locator(plt.NullLocator())
plt.gca().xaxis.set_major_locator(ticker.MaxNLocator(integer=True))
plt.ylim(1, 1e10)
plt.yscale("log")
plt.legend(ncol=2)
plt.show()
```

## 1D distributions plots

```python
mc_processes = [
    "Higgs_2018",
    # 'TTV_2018',
    "ST_NLO_2018",
    "WJets_2018",
    "VV+VVV_2018",
    "TT_powheg_2018",
    "DY_2018",
    "QCD_Pt_MuEnrichedPt5_2018",
]

signal_processes = [
    "GluGluToSUEP_mS125.000_mPhi8.000_T8.000_modeleptonic_13TeV_2018",
    "GluGluToSUEP_mS125.000_mPhi4.000_T16.000_modeleptonic_13TeV_2018",
    "GluGluToSUEP_mS125.000_mPhi8.000_T16.000_modeleptonic_13TeV_2018",
    "GluGluToSUEP_mS125.000_mPhi8.000_T32.000_modeleptonic_13TeV_2018",
]

names = [
    "muon_pt_vs_nMuon",
    "muon_ip3d_vs_nMuon",
    "muon_iso_vs_nMuon",
    "dimuon_mass_hard_vs_nMuon",
    "dimuon_mass_best_vs_nMuon",
]

nMuon_cut = 5j

for name in names:
    hists_mc = []
    hist_bkg_total = None
    for process in mc_processes:
        h_mc = plots[process][name][:, nMuon_cut::sum]
        hists_mc.append(h_mc)
        if hist_bkg_total is None:
            hist_bkg_total = h_mc.copy()
        else:
            hist_bkg_total += h_mc.copy()

    hists_signal = []
    for process in signal_processes:
        h_s = plots[process][name][:, nMuon_cut::sum]
        hists_signal.append(h_s)

    fig, ax = plt.subplots(figsize=(14, 13))

    hep.histplot(
        hists_mc,
        yerr=[np.sqrt(h.variances()) for h in hists_mc],
        label=[label.replace("_2018", "") for label in mc_processes],
        stack=True,
        histtype="fill",
        ec="black",
        lw=2,
        ax=ax,
    )

    hep.histplot(
        hists_signal,
        yerr=[np.sqrt(h.variances()) for h in hists_signal],
        label=[
            s.replace("GluGluTo", "")
            .replace(".000", "")
            .replace("mode", "")
            .replace("_13TeV_2018", "")
            for s in signal_processes
        ],
        lw=3,
        ax=ax,
    )

    hep.cms.label(llabel="Preliminary", data=True, lumi=55, ax=ax)
    plt.gca().xaxis.set_minor_locator(plt.NullLocator())
    plt.gca().xaxis.set_major_locator(ticker.MaxNLocator(integer=True))
    plt.ylim(1, 1e10)
    if ("muon_pt" not in name) and ("dimuon" not in name):
        plt.xscale("log")
    plt.yscale("log")
    plt.legend(ncol=2)
    plt.show()
```
